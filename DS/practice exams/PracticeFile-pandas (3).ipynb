```python
# importing libraries to be used
import pandas as pd
```

```python
# Create a data frame
data1={'Name': pd.Series(['Esther','Liz','John','Lin'], index=['a','b','c','d']),
       'Age': pd.Series([22,23,15,19], index=['a','b','c','d']), 'Price':pd.Series([2000,3000,4000],
      index=['a','b','d']), 'Prefered':pd.Series([True,False,True,False], index=['a','b','c','d'])}
```

Another way

```python
names=['Esther','Liz','John','Lin','Ritah','Peter']
age=[22,23,15,19,23,41]
prices=[2000,3000,4000,5000,3500,6000]
preferences=[True,False,True,False,False,False]
indexes=['a','b','c','d','e','f']
```

```python
customers_data={'Name': pd.Series(names, index=indexes), 'Age':pd.Series(age, index=indexes),
          'Price':pd.Series(prices, index=indexes),
         'Preference':pd.Series(preferences, index=indexes)}
```

```python
customers= pd.DataFrame(customers_data)
```

```python
customers
```
