<h3>Importing Required Modules </h3>

```python
import pandas as pd
```

<h4>1.Data frame Create the above table in a pandas dataframe</h4>

```python

#Creating variables to store data for each column
transaction_ids = [1,2,3,4,5]
product_names = ['Pen','Notebook','Eraser','Laptop Bag','USB Cable']
categories = ['Stationery','Stationery','Stationery','Accessories','Electronics']
prices = [500,2000,100,15000,5000]
quantities = [10,5,20,2,3]
customer_ids = [201,202,203,204,205]
years = [2024,2024,2023,2022,2023]

#Creating a dictionary for the sales data
sales_data = {
     'transaction_id': pd.Series(transaction_ids),
    'product_name':pd.Series(product_names),
    'category':pd.Series(categories),
    'price': pd.Series(prices),
    'quantity':pd.Series(quantities),
    'customer_id':pd.Series(customer_ids),
    'year': pd.Series(years)
}

#Creating the dataframe
sales_dataframe = pd.DataFrame(sales_data)

#Displaying the dataframe
sales_dataframe
```

<h3>2. Basic Inspection</h3>

<h3>What are the first three rows of the dataset</h3>

```python
#Using the head() function
sales_dataframe.head(3)
```

<h3>What are the last two rows of the dataset</h3>

```python
#Using the head() function
sales_dataframe.tail(2)
```

<h3>How many unique categories of products are there</h3>

```python
unique_categories_total = sales_dataframe['category'].nunique() #total
unique_categories = sales_dataframe['category'].unique() #products

#customized text to display the total number of unique product categories
print(f"There are {unique_categories_total} unique product categories which are: ")
#Also showing the unique product categories
for i in unique_categories:
    print(f"{i}")
```

<h4>3. Filtering and Selection</h4>

```python
#Filtering the data against the Stationery category
stationery_data = sales_dataframe[ sales_dataframe['category'] == 'Stationery']
stationery_data
```

<h4> b) Select the transactions made by customer_id 204</h4>

```python
#Filtering the data against customer id 204
customer_id204_transactions = sales_dataframe[ sales_dataframe['customer_id'] == 204]
customer_id204_transactions
```

```python
#Selecting the data where the product name is USB Cable
usb_cable = sales_dataframe[ sales_dataframe['product_name'] == 'USB Cable']
usb_cable
```

<h4> d) Select only product name, category and price</h4>

```python
#Accessing only the product name, category and price
name_category_price_data = sales_dataframe[['product_name', 'category', 'price']]
name_category_price_data
```

<h4> e) Show data where the category is stationery and price is less than 1000</h4>

```python
#Boolean indexing filter where the category is stationery and price is less than 1000
stationery_less_1000 = sales_dataframe[(sales_dataframe['category'] == 'Stationery') & (sales_dataframe['price'] < 1000)]
stationery_less_1000
```

<h4> f) Display only the below portion of the dataset</h4>

```python
#Slicing only the category, price and quantity column and getting the bottom 2 rows
accessories_or_electronics_data = sales_dataframe.iloc[-2:,2:5]
accessories_or_electronics_data
```

<h4>4. Sorting</h4>

<h4> a) Sort the dataset by price in ascending order</h4>

```python
#Sorting the data in ascending order based on price
price_sorted_data = sales_dataframe.sort_values(by='price',ascending=True)
#Displaying the sorted dataset
price_sorted_data
```

<h4> b) Sort the transactions by year in descending order</h4>

```python
#Sorting the data in descending order based on year
year_sorted_data = sales_dataframe.sort_values(by='year',ascending=False)
#Displaying the sorted dataset
year_sorted_data
```

<h4>5. Calculations</h4>

<h4>a) Create a new column called total_value</h4>

```python
#New column sales_dataframe
sales_dataframe['total_value'] =sales_dataframe['price'] * sales_dataframe['quantity']
#Displaying the updated dataframe
sales_dataframe
```

<h4>b) What is the total value of all transactions</h4>

```python
total_transactions_price = sales_dataframe['total_value'].sum()
print(f"The Total price of all transactions is:  {total_transactions_price:,}")
```
