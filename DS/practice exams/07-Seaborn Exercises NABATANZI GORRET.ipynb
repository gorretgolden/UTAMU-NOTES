```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic = sns.load_dataset('titanic')
```

# Exercises

** Recreate the plots below using the titanic dataframe. There are very few hints since most of the plots can be done with just one or two lines of code and a hint would basically give away the solution. Keep careful attention to the x and y labels for hints.**

** *Note! In order to not lose the plot image, make sure you don't code in the cell that is directly above the plot, there is an extra cell above that one which won't overwrite that plot!* **

```python
# CODE HERE6.
plt.figure(figsize=(3,3))
sns.jointplot(x='fare',y='age', data=titanic, xlim=(-100,600))
plt.grid(True)
plt.show()
```

```python
# CODE HERE
plt.figure(figsize=(4,3))
histplot = sns.histplot(data = titanic, x = 'fare', binwidth = 17, color = 'red', alpha = 0.40)
histplot.set(xlim = (0, 600))
histplot.set(ylim = (0, 500))
plt.show()
```

```python
# CODE HERE
plt.figure(figsize=(4,3))
sns.boxplot(x='class',y='age', data=titanic, palette=['#1EA5CE', '#90EFB7', '#EBB274'])
plt.show()
```

```python
# CODE HERE
plt.figure(figsize=(4,3))
sns.swarmplot(x='class',y='age', data=titanic,palette='Set2')
plt.show()
```

```python
# CODE HERE
plt.figure(figsize=(4,3))
sns.countplot(x='sex', data=titanic,palette='viridis', hue='sex',legend=False)
plt.show()
```

```python
# CODE HERE
columns = ['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare', 'adult_male', 'alone']
titanic_num = titanic[columns]
plt.figure(figsize=(4,3))
sns.heatmap(titanic_num.corr(), cmap='coolwarm')
#plt.ylim(-0.8, 0.8)
plt.show()
```

```python
# CODE HERE
plt.figure(figsize=(2,2))
histo_graph = sns.FacetGrid(data=titanic,col='sex')
histo_graph.map(plt.hist,'age')
plt.grid(True)
plt.show()
```
