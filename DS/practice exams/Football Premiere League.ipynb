```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Creating the dataframe
clubs = ['ManCity','Arsenal','Liverpoo','ArstonVilla','Totenham','NewCastle','ManUnited','WestHam','Chelsea','Brighton']
matches_played =[32,32,32,33,32,32,32,33,30,32]
wins= [22,22,21,19,18,15,15,13,12,11]
draws = [7,5,8,6,6,5,5,9,8,11]
losses = [3,5,3,8,8,12,12,11,10,10]
goals_forward = [76,75,72,68,65,69,47,52,55,52]
goals_against = [32,26,31,49,49,52,48,58,52,50]

Pleague = pd.DataFrame({
    'Club':pd.Series(clubs),
    'MP':pd.Series(matches_played),
    'W':pd.Series(wins),
    'D':pd.Series(draws),
    'L':pd.Series(losses),
    'GF':pd.Series(goals_forward),
    'GA':pd.Series(goals_against)
})
Pleague
```

<h4>Use seaborn to do any two visualizations that can explain insights from this data (the premier league 
table) (10 points)</h4>

```python
#Displaying the clubs and their points
#Adding barchart 
plt.figure(figsize=(6,4))
plt.title("A bar chart showing the club performance by their points score")
sns.barplot(x="Club",y="Pts",data=Pleague, palette='viridis')
plt.xticks(rotation=90)
plt.xlabel("Clubs")
plt.ylabel("Points")
plt.show()
```

```python
# Plot histogram of goals scored
sns.histplot(Pleague['GF'],color='blue')
plt.title('Distribution of Goals Scored (GF)')
plt.xlabel('Goals Scored')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot histogram of Matches played
sns.histplot(Pleague['MP'],color='blue')
plt.title('Distribution of Matches Played')
plt.xlabel('Goals Scored')
plt.ylabel('Frequency')
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x='GF', y='GA', hue='Club', data=Pleague)
plt.title('Goals Scored vs. Goals Allowed')
plt.xlabel('Goals Scored (GF)')
plt.ylabel('Goals Allowed (GA)')
plt.axline((0, 0), slope=1, color='red', linestyle='--', label='Equal GF and GA')
plt.legend(title='Club')
plt.tight_layout()
plt.show()
```

```python

# KDE Plot
sns.kdeplot(Pleague['GF'], shade=True, color="purple")
plt.title("KDE Plot of Goals Scored (GF)")
plt.xlabel("Goals Scored")
plt.ylabel("Density")
plt.show()
```

```python
# Line plot for points vs. matches played
plt.figure(figsize=(10, 6))
sns.lineplot(x='MP', y='Pts', hue='Club', data=Pleague)
plt.title('Points Accumulation by Matches Played')
plt.xlabel('Matches Played (MP)')
plt.ylabel('Points (Pts)')
plt.legend(title='Club', bbox_to_anchor=(1.05, 1), loc='upper)
plt.tight_layout()
plt.show()
```

```python
# Box plot for goal difference
plt.figure(figsize=(8, 6))
sns.boxplot(x='GD', y='Club', data=Pleague, palette='coolwarm', order=Pleague.sort_values('GD', ascending=False)['Club'])
plt.title('Goal Difference Distribution by Club')
plt.xlabel('Goal Difference (GD)')
plt.ylabel('Club')
plt.tight_layout()
plt.show()
```
