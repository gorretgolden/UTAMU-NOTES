<h4>b. What are some of the tasks you would ordinarily do in data wrangling?</h4>

<p>Data Wrangling is the process of transforming raw data into an organized and clean format good for data analysis. It involves the following tasks;</p>
<ul>
<li>Data Cleaning, this is a data wrangling task that involves cleaning data by removing missing values, removing duplicate entries, correcting inconsitent data formats</li>
    <li>Data Transformation, this is the process of converting data from one format or structure into another, it may also involve tasks like normalization,</li>
    <li>Data Integration,this involves combining data from multiple sources to create a unified, coherent dataset for analysis.</li>
    <li>Selecting and subsetting data is the process of narrowing down a dataset to a specific subset of records or columns relevant to the analysis</li>
</ul>

<h4>e. Why do you think data visualization is important?. Highlight a scenario where you would see data</h4>

Data visualization is the graphical representation of information and data. It plays a crucial role in understanding complex datasets and communicating insights effecttively. Simplifies Complex Data:

Raw data can be overwhelming. Visualization transforms numbers into clear and intuitive charts, graphs, and maps.
Enhances Understanding:

Visuals are processed faster by the brain than text, making it easier to grasp key insights quickly.
Reveals Trends and Patterns:

Hidden trends that might go unnoticed in raw data become apparent in visualizations.
Improves Communication:

Sharing data insights with non-technical stakeholders is easier when using visuals rather than tables or statistics alone.
Facilitates Decision-Making:

Decision-makers can act quickly and confidently when presented with clear visual representations of the data.
An e-commerce company wants to analyze its sales data to improve revenue and identify seasonal trends.

How Visualization Helps:

Line Charts: Show monthly sales trends over the year, highlighting peak seasons like Black Friday or Christmas.
Pie Charts: Visualize the contribution of different product categories to total revenue.
