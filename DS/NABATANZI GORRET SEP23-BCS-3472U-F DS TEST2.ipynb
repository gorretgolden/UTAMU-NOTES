4. What is a normal distribution ? Why is it such a big deal ?

<ul>
  <li>A normal distribution is a type of probability distribution that is symmetrical and bell-shaped, where most values tend to cluster around the mean, and fewer values appear as you move further from the center.</li>
  <li>It is defined by two key properties: the mean, which represents the average, and the standard deviation, which shows how spread out the values are.</li>
</ul>

Why it's important
<ul>
  <li>The normal distribution is commonly observed in various natural and social phenomena, such as human height, test scores, and measurement errors.</li>
  <li>Many statistical techniques, including hypothesis testing and regression analysis, rely on the assumption that data follows a normal distribution.</li>
  <li>It allows for practical predictions using the 68-95-99.7 rule, which helps estimate the probability of observing values within certain ranges.</li>
</ul>

```python
# Do some analytics
def get_number_mixedschs(data):
    # Return number of mixed schools
    mixed_schools_data  = data[data['Girls/Boys/Mixed'] == 'MIXED'] #filtering only mixed schools
    num_mixed_schools = len(mixed_schools_data) #total sum of mixed schools
    return num_mixed_schools


#printing the output by calling the function
print(f"1. The total number of mixed schools is: {get_number_mixedschs(data)}")


def get_number_gals_private(data):
    # Return number of girls only schools that are private
    #Filtering girls private schools
    gals_only_private_data = data[ (data['Girls/Boys/Mixed'] == "GIRLS ONLY") & (data['Public or Private'] == 'PRIVATE')] 
    num_gals_private = len(gals_only_private_data)
    
    return num_gals_private

#printing the output by calling the function get_number_gals_private(data)
print(f"\n2. The total number of private girls schools is: {get_number_gals_private(data)}")

def get_number_boardingonly_religious(data):
    # Return number of boarding only schools that are religiously sponsored
    #Filtering religiously sponsored boarding schools only
    boarding_religious_sponsor_data = data[ (data['Day or Boarding'] == "BOARDING ONLY") & (data['School Sponsor'] == 'RELIGIOUS ORGANIZATION')] 
    num_boarding_religious = len(boarding_religious_sponsor_data) #total length

    return num_boarding_religious

#printing the output by calling the function get_number_boardingonly_religious(data)
print(f"\n3. The total number of boarding only schools that are religiously sponsored is: {get_number_boardingonly_religious(data)}")


def get_public_mixed_school(data):
    # Return the name of the school and the total enrolment of the public, mixed school
    # with the highest number of enroled students.
    
    #First filtering the public mixed schools
    public_mixed_schools = data[ (data['Girls/Boys/Mixed'] == "MIXED") & (data['Public or Private'] == 'PUBLIC')] 
   
    #Sorting the data by enrolment in descending order
    sorted_data =  public_mixed_schools.sort_values(by='Total Enrolment 2007',ascending=False)
    school = sorted_data.iloc[0]['Name of School'] #school name
    num_enroled = sorted_data.iloc[0]['Total Enrolment 2007'] #total number of enrollments
    
    return school, num_enroled

#printing the output by calling the function get_public_mixed_school(data)
print(f"\n4. The public and mixed school with the highest number of enrolled students is: {get_public_mixed_school(data)[0]} with {get_public_mixed_school(data)[1]} enrolled students ")



def get_county_lowest_enrolment(data):
    # Return the county with the lowest enrolment number

    # Grouping the data by County and finding  the sum of  the total enrollment for each county
    county_enrollment = data.groupby('County')['Total Enrolment 2007'].sum()
    
    # Finding the county with the lowest total enrollment
    lowest_enrol_county = county_enrollment.idxmin()
    
    return lowest_enrol_county

# Getting the county with the lowest total enrollment
lowest_enrolled_county = get_county_lowest_enrolment(data)

# Getting the total enrollment for the lowest enrollment county
lowest_enrolled_number = data.groupby('County')['Total Enrolment 2007'].sum().loc[lowest_enrolled_county]

print(f"\n5. The county with the lowest enrollment number is: {lowest_enrolled_county} with a total enrollment of: {lowest_enrolled_number}")
```
